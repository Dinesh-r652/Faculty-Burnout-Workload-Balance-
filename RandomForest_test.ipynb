```python
# Import the joblib library for loading and saving Python objects efficiently.
import joblib
# Import the pandas library for creating and manipulating tabular datasets (DataFrames).
import pandas as pd
```

```python
# Define the file path where the pretrained model pipeline is stored.
model_path = "/content/faculty_burnout_model_pipeline.pkl"

# Load the pretrained machine learning model pipeline from the specified path using joblib.
loaded_model = joblib.load(model_path)

# Print a confirmation message to indicate that the model has been successfully loaded.
print("Pretrained model loaded successfully.")
```

```python
# Define a function to predict faculty burnout risk based on various input features.
def predict_burnout_risk(
    teaching_hours,
    advising_students,
    committee_count,
    research_hours,
    admin_hours,
    semester_progress,
    historical_leave_days
):
    import pandas as pd # Import pandas locally within the function to ensure 'pd' is defined.

    # Create a pandas DataFrame from the input parameters.
    # This ensures the input data is in the correct format for the model.
    input_data = pd.DataFrame([{
        "Teaching_Hours": teaching_hours,
        "Advising_Students": advising_students,
        "Committee_Count": committee_count,
        "Research_Hours": research_hours,
        "Admin_Hours": admin_hours,
        "Semester_Progress": semester_progress,
        "Historical_Leave_Days": historical_leave_days
    }])

    # Use the loaded machine learning model to make a prediction on the input data.
    # The [0] selects the first (and only) prediction result.
    prediction = loaded_model.predict(input_data)[0]

    # Get the probability estimates for each class (e.g., 'Low', 'Medium', 'High' burnout risk).
    # The [0] selects the probabilities for the first (and only) input row.
    probabilities = loaded_model.predict_proba(input_data)[0]

    # Create a dictionary to store the class names and their corresponding prediction probabilities.
    probability_dict = {
        # Iterate through class names and their probabilities.
        class_name: round(float(probability), 4) # Round probabilities to 4 decimal places.
        for class_name, probability in zip(
            loaded_model.classes_, # Get the list of class names from the model.
            probabilities # Use the calculated probabilities.
        )
    }

    # Return a dictionary containing the predicted burnout risk and the prediction probabilities for all classes.
    return {
        "Predicted_Burnout_Risk": prediction,
        "Prediction_Probabilities": probability_dict
    }
```

```python
# Define a scenario representing a faculty member with a low workload.
# Call the prediction function with specific input values for this scenario.
low_workload_case = predict_burnout_risk(
    teaching_hours=10,        # Number of teaching hours per week.
    advising_students=4,      # Number of students advised.
    committee_count=1,        # Number of committees served on.
    research_hours=15,        # Number of research hours per week.
    admin_hours=2,            # Number of administrative hours per week.
    semester_progress=0.20,   # Percentage of the semester completed (e.g., 20%).
    historical_leave_days=1   # Number of historical leave days taken.
)

# Display the prediction results for the low workload case.
low_workload_case
```

```python
# Define a scenario representing a faculty member with a medium workload.
# Call the prediction function with specific input values for this scenario.
medium_workload_case = predict_burnout_risk(
    teaching_hours=22,        # Number of teaching hours per week.
    advising_students=15,     # Number of students advised.
    committee_count=3,        # Number of committees served on.
    research_hours=8,         # Number of research hours per week.
    admin_hours=7,            # Number of administrative hours per week.
    semester_progress=0.60,   # Percentage of the semester completed (e.g., 60%).
    historical_leave_days=5   # Number of historical leave days taken.
)

# Display the prediction results for the medium workload case.
medium_workload_case
```

```python
# Define a scenario representing a faculty member with a high workload.
# Call the prediction function with specific input values for this scenario.
high_workload_case = predict_burnout_risk(
    teaching_hours=34,        # Number of teaching hours per week.
    advising_students=28,     # Number of students advised.
    committee_count=7,        # Number of committees served on.
    research_hours=3,         # Number of research hours per week.
    admin_hours=14,           # Number of administrative hours per week.
    semester_progress=0.95,   # Percentage of the semester completed (e.g., 95%).
    historical_leave_days=12  # Number of historical leave days taken.
)

# Display the prediction results for the high workload case.
high_workload_case
```

```python
# Define a list of dictionaries, where each dictionary represents a test case for burnout prediction.
# These test cases cover different workload scenarios (Low, Medium, High).
test_cases = [
    {
        "Test_Case": "Low Workload Faculty", # Descriptive name for the test case.
        "Teaching_Hours": 10,                 # Input feature: teaching hours.
        "Advising_Students": 4,              # Input feature: number of advised students.
        "Committee_Count": 1,                # Input feature: number of committees.
        "Research_Hours": 15,                # Input feature: research hours.
        "Admin_Hours": 2,                    # Input feature: administrative hours.
        "Semester_Progress": 0.20,           # Input feature: semester progress.
        "Historical_Leave_Days": 1           # Input feature: historical leave days.
    },
    {
        "Test_Case": "Medium Workload Faculty", # Descriptive name for the test case.
        "Teaching_Hours": 22,
        "Advising_Students": 15,
        "Committee_Count": 3,
        "Research_Hours": 8,
        "Admin_Hours": 7,
        "Semester_Progress": 0.60,
        "Historical_Leave_Days": 5
    },
    {
        "Test_Case": "High Workload Faculty", # Descriptive name for the test case.
        "Teaching_Hours": 34,
        "Advising_Students": 28,
        "Committee_Count": 7,
        "Research_Hours": 3,
        "Admin_Hours": 14,
        "Semester_Progress": 0.95,
        "Historical_Leave_Days": 12
    }
]
```

```python
# Initialize an empty list to store the prediction results for each test case.
results = []

# Iterate through each test case defined in the 'test_cases' list.
for case in test_cases:
    # Call the 'predict_burnout_risk' function with the input features from the current test case.
    prediction_result = predict_burnout_risk(
        teaching_hours=case["Teaching_Hours"],
        advising_students=case["Advising_Students"],
        committee_count=case["Committee_Count"],
        research_hours=case["Research_Hours"],
        admin_hours=case["Admin_Hours"],
        semester_progress=case["Semester_Progress"],
        historical_leave_days=case["Historical_Leave_Days"]
    )

    # Append a dictionary containing the test case name, predicted burnout risk, and probabilities to the 'results' list.
    results.append({
        "Test_Case": case["Test_Case"],
        "Predicted_Burnout_Risk": prediction_result[
            "Predicted_Burnout_Risk" # Extract the predicted burnout risk from the result.
        ],
        "Prediction_Probabilities": prediction_result[
            "Prediction_Probabilities" # Extract the prediction probabilities from the result.
        ]
    })

# Convert the list of results into a pandas DataFrame for better readability and analysis.
results_df = pd.DataFrame(results)

# Display the resulting DataFrame, showing the test case, predicted risk, and probabilities.
results_df
```
